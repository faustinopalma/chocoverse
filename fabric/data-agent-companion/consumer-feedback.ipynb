```python
from fabric.dataagent.client import FabricDataAgentManagement

management = FabricDataAgentManagement(
    data_agent="da_giandoria_companion_story", workspace="ws-giandoria-demo"
)
feedback = management.list_feedback(max_results=20, page_size=20)
if feedback.empty:
    raise RuntimeError(
        "No shared feedback was returned. Verify M365 publication, administrator sharing, "
        "republishing and a real consumer submission; do not mark this demo passed."
    )
print("Shared feedback rows:", len(feedback))
print(feedback["rating"].value_counts(dropna=False).to_string())
```
